### Imports and set file paths

```python
import sys # standard imports
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.mixture import GaussianMixture # custom and specialised imports
from scipy.stats import gaussian_kde
from mrs_analysis_utils import read_xml, load_sources, load_weights

s_tree_path = 'data/1TE-ShortMRS(corregido).xml'
l_tree_path = 'data/1TE-LongMRS(corregido).xml'
case_folder = 'data/split_cases/'
results_folder = 'results/cNMF/'
experiments = [['a2', 'no'], ['mm', 'me'], ['mm', 'me', 'a2'], ['mm', 'me', 'no'], ['mm', 'gl', 'a2'], ['mm', 'gl', 'no']]
colours_dict_label = {'no' : 'green', 'a2' : 'blue', 'mm' : 'darkgray', 'gl' : 'red', 'me' : 'orange'}
best_sources = {'MMvME@LTE'   : '10', 'MMvME@STE'   : '10', # select the opitimal cNMF runs for
                'A2vNO@LTE'   : '10',  'A2vNO@STE'   : '2', # each experiment
                'MMvMEvA2@LTE': '3',  'MMvMEvA2@STE': '3',
                'MMvMEvNO@LTE': '3',  'MMvMEvNO@STE': '3',
                'MMvGLvA2@LTE': '1',  'MMvGLvA2@STE': '1',
                'MMvGLvNO@LTE': '1',  'MMvGLvNO@STE': '2',
               }

debug = True
```

```python
# This cell reads the original 1TE-[Long,Short]MRS(corregido).xml, splits into individual
# cases within specific experiment subfolders within case_folder
for experiment in experiments:
    experiment_name = 'v'.join([e.upper() for e in experiment]) # 'a2', 'no' -> 'A2vNO'

    for te_type in ['L', 'S']:
        out_dir = os.path.join(case_folder, experiment_name, f'{te_type}TE')
        os.makedirs(out_dir, exist_ok=True)

        inp = s_tree_path if te_type=='S' else l_tree_path
        print(f"Saving to {out_dir}")
        read_xml(input_dir=inp, output_dir=out_dir, exp=experiment)
```

### Extract weights and sources, organise DFs

```python
run_metadata       = {}
source_matrices    = {}
reconstructed_data = {}
raw_data_frames    = {}

for experiment in experiments:
    experiment_name = 'v'.join([e.upper() for e in experiment]) 
    
    for te_type in ['L', 'S']:
        run_key    = f'{experiment_name}@{te_type}TE'
        input_dir  = os.path.join(case_folder, experiment_name, f'{te_type}TE')
        output_dir = os.path.join(results_folder, experiment_name, f'{te_type}TE')

        os.makedirs(input_dir, exist_ok=True)
        os.makedirs(output_dir, exist_ok=True)

        print(input_dir, output_dir)
        # load sources + metadata
        sources, mean_data, labels, ppm_axis, raw_signals = load_sources(
            input_dir, output_dir, [0, 4.5], 
            best_sources[run_key], experiment
        )
        if debug: print(sources.shape)

        # load weights (ONLY 2 arguments)
        weights = load_weights(
            output_dir, 
            best_sources[run_key]
        )
        if debug: print(weights.shape)

        # reconstruct spectra
        reconstructed_matrix = np.dot(weights, sources)
        reconstructed_df     = pd.DataFrame(reconstructed_matrix)
        raw_df               = pd.DataFrame(raw_signals)

        reconstructed_df['labels'] = labels
        raw_df['labels']           = labels

        run_metadata[run_key] = {
            'sources'     : sources,
            'mean_signal' : mean_data,
            'labels'      : labels,
            'ppm_axis'    : ppm_axis,
            'weights'     : weights
        }

        source_matrices[run_key]    = sources
        reconstructed_data[run_key] = reconstructed_df
        raw_data_frames[run_key]    = raw_df
        
run_summary_df = pd.DataFrame(run_metadata)

assert reconstructed_data['A2vNO@LTE'].shape == raw_data_frames['A2vNO@LTE'].shape
```

```python
real_weights = {}

for experiment in experiments:
    experiment_name = "v".join(e.upper() for e in experiment)

    for te_type in ['L', 'S']:
        run_key = f"{experiment_name}@{te_type}TE"
        run_data = run_summary_df[run_key]

        weights_matrix = run_data['weights']
        labels = run_data['labels']

        # Build a DataFrame where each column is a weight (w1, w2, ...) and a true_label
        weight_rows = []
        for weight_vector, label in zip(weights_matrix, labels):
            row = {'true_label': label}
            # Assign weights dynamically: w1, w2, w3, ...
            for i, w in enumerate(weight_vector, start=1):
                row[f'w{i}'] = w
            weight_rows.append(row)

        # Store as DataFrame
        real_weights[run_key] = pd.DataFrame(weight_rows)

if debug: print(real_weights)
```

```python
import matplotlib.pyplot as plt

def plot_reconstructed_all(reconstructed_data, experiments):
    for experiment in experiments:

        experiment_name = 'v'.join([e.upper() for e in experiment])

        fig, axs = plt.subplots(1, 2, figsize=(14, 5), sharey=True)
        fig.suptitle(f"Reconstructed spectra for {experiment_name}", fontsize=18)

        for idx, te_type in enumerate(['L', 'S']):
            run_key = f"{experiment_name}@{te_type}TE"
            
            if run_key not in reconstructed_data:
                print(f"Missing: {run_key}")
                continue

            df = reconstructed_data[run_key]

            spectra_only = df.drop(columns=['labels']).values
            labels = df['labels'].values

            ax = axs[idx]
            x = np.arange(spectra_only.shape[1])  # if you don't have ppm_axis here

            for row, lab in zip(spectra_only, labels):
                ax.plot(x, row, alpha=0.4, label=str(lab))

            ax.set_title(f"{te_type}TE")
            ax.set_xlabel("Index")
            ax.set_ylabel("Intensity")
            
            # Optional: show unique labels in legend
            handles, lbls = ax.get_legend_handles_labels()
            uniq = dict(zip(lbls, handles))
            ax.legend(uniq.values(), uniq.keys())

        plt.tight_layout()
        plt.show()


# ---- RUN THE PLOTTER ----
plot_reconstructed_all(reconstructed_data, experiments)
```

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import seaborn as sns
import numpy as np
import pandas as pd

def plot_real_weights_and_spectra(run_key, reconstructed_data, run_metadata, colours_dict_label=None):
    """
    Plot real cNMF weights + reconstructed spectra exactly like synthetic plotting function.
    """

    # -----------------------
    #  Extract data
    # -----------------------
    df_recon = reconstructed_data[run_key].copy()
    labels = df_recon['labels'].copy()
    df_recon = df_recon.drop(columns=['labels'])

    weights = pd.DataFrame(run_metadata[run_key]["weights"]).copy()
    n_sources = weights.shape[1]

    # name columns as w1, w2, w3...
    weights.columns = [f"w{i+1}" for i in range(n_sources)]
    weights["label"] = labels.values

    # colour dictionary
    unique_labels = sorted(weights['label'].unique())

    if colours_dict_label is None:
        default_cols = sns.color_palette("tab10", len(unique_labels))
        colours_dict_label = {l: default_cols[i] for i, l in enumerate(unique_labels)}

    label_to_colour = {label: colours_dict_label.get(label, "black") for label in unique_labels}

    # -----------------------
    #  Figure layout
    # -----------------------
    fig = plt.figure(figsize=(6 * (1 + len(unique_labels)), 6))

    # 3D scatter if we have >= 3 sources
    if n_sources >= 3:
        ax0 = fig.add_subplot(1, 1 + len(unique_labels), 1, projection='3d')
        for lab in unique_labels:
            subset = weights[weights['label'] == lab]
            ax0.scatter(subset["w1"], subset["w2"], subset["w3"],
                        label=lab, color=label_to_colour[lab])
        ax0.set_zlabel("w3")
    else:
        ax0 = fig.add_subplot(1, 1 + len(unique_labels), 1)
        sns.scatterplot(data=weights,
                        x="w1", y="w2",
                        hue="label", palette=label_to_colour,
                        ax=ax0)

    ax0.set_title(f"Weights – {run_key}")
    ax0.set_xlabel("w1")
    ax0.set_ylabel("w2")
    ax0.legend()
    ax0.grid(True)

    # -----------------------
    #  Plot reconstructed spectra per label
    # -----------------------
    for i, lab in enumerate(unique_labels):
        ax = fig.add_subplot(1, 1 + len(unique_labels), i + 2)
        indices = np.where(labels == lab)[0]

        for j in indices:
            ax.plot(df_recon.iloc[j], alpha=0.6, color=label_to_colour[lab])

        ax.set_title(f"{lab} Spectra")
        ax.set_xlabel("Index (or ppm axis)")
        ax.set_ylabel("Intensity")
        ax.grid(True)

    fig.suptitle(f"Real Data – {run_key}", fontsize=16)
    plt.tight_layout()
    plt.show()
```

```python
for experiment in experiments:
    experiment_name = 'v'.join([e.upper() for e in experiment])
    for te_type in ['L', 'S']:
        key = f"{experiment_name}@{te_type}TE"
        plot_real_weights_and_spectra(key, reconstructed_data, run_metadata)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde
from matplotlib.colors import ListedColormap
from scipy.spatial.distance import mahalanobis
from mpl_toolkits.mplot3d import Axes3D
from mpl_toolkits.mplot3d import proj3d

def compute_mahalanobis_outliers(df, cols, threshold=9.0):
    data = df[cols].values
    mean = np.mean(data, axis=0)
    cov = np.cov(data, rowvar=False)
    cov_inv = np.linalg.pinv(cov)

    d2 = np.array([
        mahalanobis(row, mean, cov_inv)**2
        for row in data
    ])
    return d2 > threshold, d2


def place_outlier_labels(ax, outlier_points, title):
    """
    Places outlier labels neatly at the right-hand side margin.
    Works for 2D axes or 2D projection of 3D axes.
    """
    xlims = ax.get_xlim()
    ylims = ax.get_ylim()
    margin_x = xlims[1] + 0.05*(xlims[1]-xlims[0])    
    ys = np.linspace(ylims[1], ylims[0], len(outlier_points) + 2)

    for (label, x, y, d2), y_text in zip(outlier_points, ys):
        text_str = f"{label} (D²={d2:.1f})"
        if hasattr(ax, "zaxis"):  # 3D axes
            # Use z=0 for projected 2D labels
            ax.text(margin_x, y_text, 0, text_str, color='red', fontsize=9, va='center')
        else:  # 2D axes
            ax.text(margin_x, y_text, text_str, color='red', fontsize=9, va='center')

    ax.set_xlim(xlims[0], margin_x + 0.1*(xlims[1]-xlims[0]))

def compute_global_axes_limits(real_weights):
    x_min, x_max = np.inf, -np.inf
    y_min, y_max = np.inf, -np.inf
    z_min, z_max = np.inf, -np.inf

    for df in real_weights.values():
        df = df.copy()
        if 'w1' in df.columns:
            x_min = min(x_min, df['w1'].min())
            x_max = max(x_max, df['w1'].max())
        if 'w2' in df.columns:
            y_min = min(y_min, df['w2'].min())
            y_max = max(y_max, df['w2'].max())
        if 'w3' in df.columns:
            z_min = min(z_min, df['w3'].min())
            z_max = max(z_max, df['w3'].max())

    # Add a small buffer (5%)
    x_buffer = (x_max - x_min) * 0.05
    y_buffer = (y_max - y_min) * 0.05
    z_buffer = (z_max - z_min) * 0.05

    xlims = (x_min - x_buffer, x_max + x_buffer)
    ylims = (y_min - y_buffer, y_max + y_buffer)
    zlims = (z_min - z_buffer, z_max + z_buffer) if z_min < np.inf else None

    return xlims, ylims, zlims

def plot_weights_with_kde_and_outliers(real_weights, colours_dict_label):
    for key, df in real_weights.items():
        df = df.copy()

        print(f"Plotting {key}")

        # ----------------------------------------------------
        # CASE 1: 2D (w1, w2)
        # ----------------------------------------------------
        if set(['w1','w2']).issubset(df.columns) and 'w3' not in df.columns:
            unique_labels = sorted(df['true_label'].unique())
            fig, ax = plt.subplots(figsize=(10,10))

            all_outlier_points = []

            for label in unique_labels:
                sub = df[df['true_label']==label].copy()
                x = sub['w1'].values
                y = sub['w2'].values
                color = colours_dict_label.get(label, 'black')

                # --- Outliers ---
                out_mask, d2 = compute_mahalanobis_outliers(sub, ['w1','w2'])
                sub['is_outlier'] = out_mask
                sub['d2'] = d2

                # --- KDE background ---
                try:
                    kde = gaussian_kde(np.vstack([x,y]))
                    Xg = np.linspace(x.min(), x.max(), 250)
                    Yg = np.linspace(y.min(), y.max(), 250)
                    X, Y = np.meshgrid(Xg, Yg)
                    Z = kde(np.vstack([X.ravel(), Y.ravel()])).reshape(X.shape)
                    Znorm = (Z - Z.mean()) / Z.std()
                    levels = np.linspace(Znorm.min(), Znorm.max(), 8)

                    for i in range(len(levels)-1):
                        ax.contourf(
                            X, Y, Znorm,
                            levels=[levels[i], levels[i+1]],
                            colors=[color],
                            alpha=(i+1)/len(levels)*0.25
                        )
                except:
                    pass

                # --- Scatter normal points ---
                ax.scatter(x, y, s=20, color=color, alpha=0.85)

                # --- Collect outliers for margin labels ---
                for i, row in sub[sub['is_outlier']].iterrows():
                    all_outlier_points.append(
                        (f"OUT {label}", row['w1'], row['w2'], row['d2'])
                    )
                    ax.scatter(row['w1'], row['w2'], s=100,
                               facecolors='none', edgecolors='red', linewidths=2)

            # --- Neat margin labels ---
            place_outlier_labels(ax, all_outlier_points, key)

            ax.set_title(f"2D KDE + Outliers – {key}")
            xlims, ylims, zlims = compute_global_axes_limits(real_weights)

            # Example usage in 2D:
            ax.set_xlim(xlims)
            ax.set_ylim(ylims)
            
            # Example usage in 3D:
            ax.set_xlim(xlims)
            ax.set_ylim(ylims)
            ax.set_xlabel("w1")
            ax.set_ylabel("w2")
            ax.grid(True)
            plt.tight_layout()
            plt.show()


        # ----------------------------------------------------
        # CASE 2: 3D (w1, w2, w3)
        # ----------------------------------------------------
        if set(['w1','w2','w3']).issubset(df.columns):
            unique_labels = sorted(df['true_label'].unique())
            fig = plt.figure(figsize=(10,10))
            ax = fig.add_subplot(111, projection='3d')

            all_outlier_points = []

            for label in unique_labels:
                sub = df[df['true_label']==label].copy()
                x = sub['w1'].values
                y = sub['w2'].values
                z = sub['w3'].values
                color = colours_dict_label.get(label, 'black')

                # --- Outliers (3D mahalanobis) ---
                out_mask, d2 = compute_mahalanobis_outliers(sub, ['w1','w2','w3'])
                sub['is_outlier'] = out_mask
                sub['d2'] = d2

                # --- Scatter ---
                ax.scatter(x, y, z, s=18, color=color, alpha=0.9)

                # --- Mark outliers ---
                for i, row in sub[sub['is_outlier']].iterrows():
                    ax.scatter(row['w1'], row['w2'], row['w3'],
                               s=120, facecolors='none',
                               edgecolors='red', linewidths=2)
                    # Store projected coordinates for side labels
                    all_outlier_points.append(
                        (f"OUT {label}", row['w1'], row['w2'], row['d2'])
                    )

            # --- Use 2D projection for neat labels ---
            x2d, y2d, _ = proj3d.proj_transform(
                [p[1] for p in all_outlier_points],
                [p[2] for p in all_outlier_points],
                [0]*len(all_outlier_points),
                ax.get_proj()
            )
            projected = [(lbl, x2d[i], y2d[i], d2) for i, (lbl,_,_,d2) in enumerate(all_outlier_points)]
            place_outlier_labels(ax, projected, key)

            ax.set_title(f"3D Weights + Outliers – {key}")
            xlims, ylims, zlims = compute_global_axes_limits(real_weights)

            # Example usage in 2D:
            ax.set_xlim(xlims)
            ax.set_ylim(ylims)
            
            # Example usage in 3D:
            ax.set_xlim(xlims)
            ax.set_ylim(ylims)
            if zlims is not None:
                ax.set_zlim(zlims)
            ax.set_xlabel("w1")
            ax.set_ylabel("w2")
            ax.set_zlabel("w3")
            plt.tight_layout()
            plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import gaussian_kde, zscore
import re

def get_weight_columns(df):
    # numeric only
    numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
    # match w1, w2, w3, w4...
    weight_cols = [c for c in numeric_cols if re.match(r'^w\d+$', c)]
    return weight_cols

def plot_weights_with_outliers(real_weights):
    for key, df in real_weights.items():
        df = df.copy()

        cols = get_weight_columns(df)

        if len(cols) == 2:
            w1, w2 = cols
            data = df[[w1, w2]].dropna()

            # OUTLIERS
            zs1 = zscore(data[w1])
            zs2 = zscore(data[w2])
            outliers = (abs(zs1) > 3) | (abs(zs2) > 3)

            outlier_points = data[outliers]
            inlier_points  = data[~outliers]

            plt.figure(figsize=(10, 8))

            # KDE shading
            kde = gaussian_kde(data.T)
            xgrid = np.linspace(data[w1].min(), data[w1].max(), 200)
            ygrid = np.linspace(data[w2].min(), data[w2].max(), 200)
            X, Y = np.meshgrid(xgrid, ygrid)
            Z = kde(np.vstack([X.ravel(), Y.ravel()])).reshape(X.shape)
            plt.contourf(X, Y, Z, levels=12, alpha=0.4)

            # scatter
            plt.scatter(inlier_points[w1], inlier_points[w2], s=20, label="Inliers")
            plt.scatter(outlier_points[w1], outlier_points[w2], s=50, c='red', label="Outliers")

            # TEXT LABELS neatly aligned
            if len(outlier_points) > 0:
                x_text = data[w1].max() + (0.05 * data[w1].std())
                y_start = data[w2].min()
                y_step = (data[w2].max() - data[w2].min()) / max(len(outlier_points), 5)

                for i, (idx, row) in enumerate(outlier_points.iterrows()):
                    plt.text(x_text, y_start + i*y_step,
                             f"outlier idx {idx}: ({row[w1]:.2f},{row[w2]:.2f})",
                             fontsize=9, color="red")

            plt.title(f"{key} — 2D weights")
            plt.xlabel(w1)
            plt.ylabel(w2)
            plt.legend()
            plt.tight_layout()
            plt.show()

        elif len(cols) == 3:
            w1, w2, w3 = cols
            data = df[[w1, w2, w3]].dropna()

            zs = np.abs(zscore(data))
            outliers = (zs > 3).any(axis=1)

            outlier_points = data[outliers]
            inlier_points  = data[~outliers]

            fig = plt.figure(figsize=(10, 8))
            ax = fig.add_subplot(111, projection='3d')

            # Scatter
            ax.scatter(inlier_points[w1], inlier_points[w2], inlier_points[w3], s=20)
            ax.scatter(outlier_points[w1], outlier_points[w2], outlier_points[w3], s=60, c='red')

            # Text labels placed at the right margin
            if len(outlier_points) > 0:
                x_text = data[w1].max() + (0.1 * data[w1].std())
                for i, (idx, row) in enumerate(outlier_points.iterrows()):
                    ax.text(x_text, row[w2], row[w3],
                            f"idx {idx}",
                            color="red", fontsize=8)

            ax.set_title(f"{key} — 3D weights")
            ax.set_xlabel(w1)
            ax.set_ylabel(w2)
            ax.set_zlabel(w3)
            plt.tight_layout()
            plt.show()

        else:
            print(f"Skipping {key}: found {len(cols)} weight columns → {cols}")
```

```python
plot_weights_with_outliers(real_weights)
```

### Plotting of real weights and spectra

```python
def plot_spectra_and_weights(original_df, reconstructed_df, title, palette=colours_dict_label):
    """
    Plots original spectra, reconstructed spectra, and the weight scatterplots.
    """    
    is_2d = 'w3' not in original_df.columns
    fig = plt.figure(figsize=(18, 5))

    # Subplot 1: Original spectra
    ax0 = fig.add_subplot(131)
    for label in original_df[label_col].unique():
        subset = original_df[original_df[label_col] == label]
        for i in range(len(subset)):
            ax0.plot(subset.columns.difference([label_col]), subset.iloc[i].drop(label_col),
                     color=palette[label], alpha=0.5)
    ax0.set_title('Original Spectra', fontsize=16)
    ax0.set_xlabel('Features', fontsize=14)
    ax0.set_ylabel('Intensity', fontsize=14)

    # Subplot 2: Reconstructed spectra
    ax1 = fig.add_subplot(132)
    for label in reconstructed_df[label_col].unique():
        subset = reconstructed_df[reconstructed_df[label_col] == label]
        for i in range(len(subset)):
            ax1.plot(subset.columns.difference([label_col]), subset.iloc[i].drop(label_col),
                     color=palette[label], alpha=0.5)
    ax1.set_title('Reconstructed Spectra', fontsize=16)
    ax1.set_xlabel('Features', fontsize=14)
    ax1.set_ylabel('Intensity', fontsize=14)

    # Subplot 3: Weight scatterplot
    ax2 = fig.add_subplot(133) if is_2d else fig.add_subplot(133, projection='3d')
    if is_2d:
        sns.scatterplot(data=original_df, x='w1', y='w2', hue=label_col, palette=palette, ax=ax2)
        ax2.set_xlabel('Weight 1', fontsize=14)
        ax2.set_ylabel('Weight 2', fontsize=14)
    else:
        colours = original_df[label_col].map(palette)
        ax2.scatter(original_df['w1'], original_df['w2'], original_df['w3'], c=colours)
        ax2.set_xlabel('Weight 1', fontsize=14)
        ax2.set_ylabel('Weight 2', fontsize=14)
        ax2.set_zlabel('Weight 3', fontsize=14)
        handles = [
            plt.Line2D([0], [0], marker='o', color='w', markerfacecolor=palette[label],
                       markersize=10, label=label)
            for label in original_df[label_col].unique()
        ]
        ax2.legend(handles=handles, title='Label', loc='upper left')
    ax2.set_title('Weight Scatterplot', fontsize=16)

    fig.suptitle(title, fontsize=18)
    plt.tight_layout()
    plt.show()
```

```python
def generate_synthetic_weights(real_data, model):
    m_factor = 2
    synthetic_weights = {}
    eval_metrics = {}

    for key, weights in real_data.items():
        is2d = weights.shape[1] == 3
        temp_synthetic_weights = []

        weights_data = weights[['w1', 'w2', 'true_label']] if is2d else weights[['w1', 'w2', 'w3', 'true_label']]
        log_l_list = []

        for label in weights_data['true_label'].unique():
            group = weights_data[weights_data['true_label'] == label][['w1', 'w2']] if is2d else \
                    weights_data[weights_data['true_label'] == label][['w1', 'w2', 'w3']]

            group = group.values
            try:
                if group.shape[0] < model.n_components:
                    continue
            except:
                pass

            num_cases = group.shape[0]
            model.fit(group)

            # Sample until we have enough strictly positive samples
            accepted = []
            while len(accepted) < num_cases * m_factor:
                needed = num_cases * m_factor - len(accepted)
                sampled = model.sample(needed * 2)[0] if isinstance(model, GaussianMixture) else model.sample(needed * 2)

                strictly_positive = sampled[(sampled > 0).all(axis=1)]
                accepted.extend(strictly_positive[:needed])

            synth_weight = np.array(accepted[:num_cases * m_factor])
            temp_data_pt = pd.DataFrame(synth_weight, columns=['w1', 'w2'] if is2d else ['w1', 'w2', 'w3'])
            temp_data_pt['true_label'] = label
            temp_synthetic_weights.append(temp_data_pt)

            if not isinstance(model, GaussianMixture):
                log_l = model.score_samples(group)
                log_l_list.extend(log_l)

        synthetic_weights[key] = pd.concat(temp_synthetic_weights, ignore_index=True)

        if isinstance(model, GaussianMixture):
            eval_metrics[key] = f'{model.bic(group):.4f}'
        else:
            n = len(group)
            k = group.shape[1]
            ll = np.sum(log_l_list)
            aic = 2 * k - 2 * ll
            bic = np.log(n) * k - 2 * ll
            eval_metrics[key] = f'{bic:.4f}'

    return synthetic_weights, eval_metrics
```

```python
gmm_evals = {}
gmm_synthetic_weights = {}

for n_comps in range(1, 11):
    synthetic_weights, evals = generate_synthetic_weights(
        real_weights, 
        GaussianMixture(n_components=n_comps, covariance_type='full')
    )

    for key in synthetic_weights:
        if key not in gmm_synthetic_weights:
            gmm_synthetic_weights[key] = {}
    
    for key in evals:
        if key not in gmm_evals:
            gmm_evals[key] = {}

        gmm_synthetic_weights[key][n_comps] = synthetic_weights[key]
        gmm_evals[key][n_comps] = evals[key]

# Create and display the evaluation results DataFrame
gmm_evals_df = pd.DataFrame(gmm_evals).reindex(range(1, 11))
gmm_evals_df = gmm_evals_df.applymap(lambda x: np.nan if isinstance(x, str) and x.replace('.', '', 1).isdigit() and float(x) > 0 else x)
gmm_evals_df
```

```python
from sklearn.neighbors import KernelDensity

bandwidths = np.logspace(0, -4, 15)
kernels    = ['gaussian', 'tophat']
kde_evals = {}
kde_synthetic_weights = {}

for bw in bandwidths:
    for krnl in kernels:
        synthetic_weights, evals = generate_synthetic_weights(
            real_weights, 
            KernelDensity(kernel=krnl, bandwidth=bw)
        )
        
        for key in synthetic_weights:
            if key not in kde_synthetic_weights:
                kde_synthetic_weights[key] = {}
                
        for key in evals:
            if key not in kde_evals:
                kde_evals[key] = {}

            kde_synthetic_weights[key][(bw, krnl)] = synthetic_weights[key]
            kde_evals[key][(bw, krnl)] = evals[key]

# Create MultiIndex for (Kernel, Bandwidth) combinations
rows = pd.MultiIndex.from_tuples(
    [(krnl, f'bw={bw:.4f}') for bw in bandwidths for krnl in kernels],
    names=['Kernel', 'Bandwidth']
)

# Build and sort DataFrame
kde_eval_df = pd.DataFrame(kde_evals)
kde_eval_df.index = rows
kde_eval_df = kde_eval_df.sort_index()

# Display the result
kde_eval_df
```
