```python
def shade_by_distance(base_color, distance, dmin, dmax):
    """
    Produce clearly distinguishable light/dark shades without blowing out colour.
    """
    # normalise distance to [0, 1]
    if dmax == dmin:
        t = 0.5
    else:
        t = (distance - dmin) / (dmax - dmin)

    # Convert to HSV
    r, g, b = to_rgb(base_color)
    h, s, v = rgb_to_hsv([r, g, b])

    # -------------------------------
    # Controlled brightness range
    # -------------------------------
    v_dark  = 0.10
    v_light = 0.95
    v_new   = v_dark + (v_light - v_dark) * (1 - t)  # closer → lighter

    # -------------------------------
    # Small saturation correction
    # -------------------------------
    # Further points (dark) get slightly more saturation (pops more)
    # Closer points (light) get slightly less saturation
    s_min = max(0, s - 0.15)
    s_max = min(1, s + 0.10)
    s_new = s_max - (s_max - s_min) * (1 - t)

    # Convert back
    r2, g2, b2 = hsv_to_rgb([h, s_new, v_new])
    return (r2, g2, b2)


def plot_real_weights_and_spectra(run_key, reconstructed_data, run_metadata, colours_dict_label=None):
    """
    Plot real cNMF weights + reconstructed spectra with brightness varying
    by distance from origin.
    """

    # -----------------------
    #  Extract data
    # -----------------------
    df_recon = reconstructed_data[run_key].copy()
    labels = df_recon['labels'].copy()
    df_recon = df_recon.drop(columns=['labels'])

    weights = pd.DataFrame(run_metadata[run_key]["weights"]).copy()
    n_sources = weights.shape[1]
    weights.columns = [f"w{i+1}" for i in range(n_sources)]
    weights["label"] = labels.values

    # base label colours
    unique_labels = sorted(weights['label'].unique())
    label_to_colour = {lab: colours_dict_label.get(lab, "black") for lab in unique_labels}

    # -----------------------
    #  Distance → brightness
    # -----------------------
    w_vals = weights[[f"w{i+1}" for i in range(n_sources)]].values
    distances = np.linalg.norm(w_vals, axis=1)   # Euclidean distance from origin
    dmin, dmax = distances.min(), distances.max()

    # Create per-point colours
    per_point_colours = []
    for idx, row in weights.iterrows():
        base = label_to_colour[row["label"]]
        dist = distances[idx]
        shade = shade_by_distance(base, dist, dmin, dmax)
        per_point_colours.append(shade)

    weights["shade"] = per_point_colours

    # -----------------------
    #  Figure layout
    # -----------------------
    fig = plt.figure(figsize=(6 * (1 + len(unique_labels)), 6))

    # 3D scatter if >=3 sources
    if n_sources >= 3:
        ax0 = fig.add_subplot(1, 1 + len(unique_labels), 1, projection='3d')
        for lab in unique_labels:
            subset = weights[weights["label"] == lab]
            ax0.scatter(
                subset["w1"], subset["w2"], subset["w3"],
                color=list(subset["shade"].values),
                label=lab
            )
        ax0.set_zlabel("w3")

    else:
        ax0 = fig.add_subplot(1, 1 + len(unique_labels), 1)
        sns.scatterplot(
            data=weights,
            x="w1", y="w2",
            hue="label",  # still group by label for legend
            palette=label_to_colour,
            ax=ax0,
            legend=False  # we handle legend manually
        )
        # overlay shaded points
        ax0.scatter(
            weights["w1"], weights["w2"],
            c=list(weights["shade"].values)
        )

    ax0.set_title(f"Weights – {run_key}")
    ax0.set_xlabel("w1")
    ax0.set_ylabel("w2")
    ax0.legend()
    ax0.grid(True)

    # -----------------------
    #  Plot reconstructed spectra per label, using shaded colour
    # -----------------------
    for i, lab in enumerate(unique_labels):
        ax = fig.add_subplot(1, 1 + len(unique_labels), i + 2)

        subset_idx = np.where(labels == lab)[0]
        shades_subset = weights.loc[weights["label"] == lab, "shade"].values

        for j, col in zip(subset_idx, shades_subset):
            ax.plot(df_recon.iloc[j], color=col, alpha=0.9)

        ax.set_title(f"{lab} Spectra")
        ax.set_xlabel("Index (or ppm axis)")
        ax.set_ylabel("Intensity")
        ax.grid(True)

    fig.suptitle(f"Real Data – {run_key}", fontsize=16)
    plt.tight_layout()
    #plt.show()
    plt.close(fig)

for experiment in experiments:
    experiment_name = 'v'.join([e.upper() for e in experiment])
    for te_type in ['L', 'S']:
        key = f"{experiment_name}@{te_type}TE"
        plot_real_weights_and_spectra(key, reconstructed_data, run_metadata, colours_dict_label)
```

```python
def compute_global_axes_limits(real_weights):
    x_min, x_max = np.inf, -np.inf
    y_min, y_max = np.inf, -np.inf
    z_min, z_max = np.inf, -np.inf
    
    for df in real_weights.values():
        if "w1" in df: x_min = min(x_min, df["w1"].min()); x_max = max(x_max, df["w1"].max())
        if "w2" in df: y_min = min(y_min, df["w2"].min()); y_max = max(y_max, df["w2"].max())
        if "w3" in df: z_min = min(z_min, df["w3"].min()); z_max = max(z_max, df["w3"].max())

    x_buffer = (x_max - x_min) * 0.05 if x_max > x_min else 0.1
    y_buffer = (y_max - y_min) * 0.05 if y_max > y_min else 0.1
    z_buffer = (z_max - z_min) * 0.05 if z_max > z_min else 0.1

    xlims = (x_min - x_buffer, x_max + x_buffer)
    ylims = (y_min - y_buffer, y_max + y_buffer)
    zlims = (z_min - z_buffer, z_max + z_buffer) if z_min < np.inf else None

    return xlims, ylims, zlims


# ---------------------------------------------------------
# 3D Gaussian ellipsoid (analytic, stable)
# ---------------------------------------------------------
def plot_gaussian_ellipsoid(ax, mean, cov, n_std=1.0, color="C0", alpha=0.15, n_points=40):
    vals, vecs = eigh(cov)
    radii = n_std * np.sqrt(np.maximum(vals, 0))

    u = np.linspace(0, 2*np.pi, n_points)
    v = np.linspace(0, np.pi, n_points)

    x = np.outer(np.cos(u), np.sin(v))
    y = np.outer(np.sin(u), np.sin(v))
    z = np.outer(np.ones_like(u), np.cos(v))

    xyz = np.stack((x, y, z), axis=-1)
    xyz *= radii
    xyz = xyz @ vecs.T
    xyz += mean

    X, Y, Z = xyz[...,0], xyz[...,1], xyz[...,2]
    ax.plot_surface(X, Y, Z, rstride=1, cstride=1,
                    color=color, alpha=alpha,
                    linewidth=0, shade=True)


# ---------------------------------------------------------
# MAIN PLOTTING FUNCTION (2D + 3D)
# ---------------------------------------------------------
def plot_weights_with_kde_and_outliers(real_weights, colours_dict_label, outlier_sd=3.0, gridsize=60):

    xlims, ylims, zlims = compute_global_axes_limits(real_weights)

    for key, df in real_weights.items():
        df = df.copy()
        print("Plotting", key)

        if "true_label" not in df and "label" in df:
            df["true_label"] = df["label"]

        # -----------------------------------------------------
        # 2D PLOT
        # -----------------------------------------------------
        if set(["w1","w2"]).issubset(df.columns) and "w3" not in df.columns:
            fig, ax = plt.subplots(figsize=(10,10), dpi=100)
            all_out = []
            unique_labels = sorted(df["true_label"].unique())

            for label in unique_labels:
                sub = df[df["true_label"] == label].copy()
                if sub.empty: continue

                # z-score outlier detection
                zs = np.abs(sub[["w1","w2"]].apply(zscore))
                out_mask = (zs > outlier_sd).any(axis=1)
                sub["is_outlier"] = out_mask

                color = colours_dict_label.get(label, "C0")
                data = sub[["w1","w2"]].astype(float).to_numpy()

                # --- Gaussian SD rings ---
                if len(sub) > 3:
                    mean = data.mean(axis=0)
                    cov = np.cov(data, rowvar=False)
                    cov += np.eye(2) * 1e-6 # THIS LINE IS SO IMPORTANT AND I DONT KNOW WHY, IF REMOVED IT CAUSES THE KERNEL PLOTTING TO BE FLAT AND CONTAIN ISLANDS
                    xg = np.linspace(*xlims, gridsize)
                    yg = np.linspace(*ylims, gridsize)
                    X, Y = np.meshgrid(xg, yg)
                    pts = np.column_stack([X.ravel(), Y.ravel()])

                    diff = pts - mean
                    cov_inv = np.linalg.pinv(cov + 1e-8*np.eye(2))
                    d2 = np.einsum("ij,jk,ik->i", diff, cov_inv, diff).reshape(X.shape)
                    D = np.sqrt(d2)

                    levels = [0,1,2,3]
                    alphas = [0.50, 0.30, 0.15]

                    for i in range(3):
                        col = np.array(mcolors.to_rgba(color))
                        col[3] = alphas[i]
                        ax.contourf(X, Y, D, levels=[i, i+1], colors=[col], antialiased=True)

                    ax.contour(X, Y, D, levels=[1,2,3], colors=[color], linewidths=1.2)

                # --- INLIERS ---
                ax.scatter(sub.loc[~out_mask, "w1"],
                           sub.loc[~out_mask, "w2"],
                           s=40, color=color, alpha=0.9, label=label, zorder=5)

                # --- OUTLIERS ---
                for _, row in sub[sub["is_outlier"]].iterrows():
                    all_out.append((label, row["w1"], row["w2"]))

                    ax.scatter(row["w1"], row["w2"],
                               s=160, facecolors="none", edgecolors="red",
                               linewidths=2, zorder=3)
                    ax.scatter(row["w1"], row["w2"],
                               s=40, color=color, edgecolor="black",
                               linewidth=0.4, zorder=6)

            print("Outliers for", key)
            for l,x,y in all_out:
                print(f"  {l}: x={x:.3f}, y={y:.3f}")

            # Legend
            handles, labels = ax.get_legend_handles_labels()
            uniq = {}
            for h,l in zip(handles, labels):
                if l not in uniq: uniq[l] = h
            handles = list(uniq.values())
            labels  = list(uniq.keys())

            out_h = mlines.Line2D([],[], marker='o', color='none',
                                  markerfacecolor='none', markeredgecolor='red',
                                  markersize=10)
            handles.append(out_h)
            labels.append(f"Outliers (n={len(all_out)})")

            ax.legend(handles, labels, title="Label")
            ax.set_xlim(xlims); ax.set_ylim(ylims)
            ax.set_xlabel("w1"); ax.set_ylabel("w2")
            ax.set_title(f"2D Gaussian SD + Outliers – {key}")
            ax.grid(True)
            plt.tight_layout()
            plt.close(fig)
            #plt.show()

        # -----------------------------------------------------
        # 3D PLOT
        # -----------------------------------------------------
        if set(["w1","w2","w3"]).issubset(df.columns):
            fig = plt.figure(figsize=(10,10), dpi=100)
            ax = fig.add_subplot(111, projection='3d')
            all_out = []
            unique_labels = sorted(df["true_label"].unique())

            for label in unique_labels:
                sub = df[df["true_label"] == label].copy()
                if sub.empty: continue
                color = colours_dict_label.get(label, "C0")

                # z-score outlier detection
                zs = np.abs(sub[["w1","w2","w3"]].apply(zscore))
                out_mask = (zs > outlier_sd).any(axis=1)
                sub["is_outlier"] = out_mask

                # Fit Gaussian to INLIERS only
                inliers = sub.loc[~out_mask, ["w1","w2","w3"]].astype(float).to_numpy()

                if len(inliers) >= 5:
                    mean3 = inliers.mean(axis=0)
                    cov3  = np.cov(inliers, rowvar=False)

                    # Draw ellipsoids (1σ, 2σ, 3σ)
                    for n_std, alpha in zip([1,2,3], [0.30, 0.18, 0.10]):
                        plot_gaussian_ellipsoid(ax, mean3, cov3,
                                                n_std=n_std, color=color,
                                                alpha=alpha)

                # Inliers
                ax.scatter(sub.loc[~out_mask,"w1"],
                           sub.loc[~out_mask,"w2"],
                           sub.loc[~out_mask,"w3"],
                           s=40, color=color, alpha=0.9,
                           label=label, zorder=5)

                # Outliers
                for _, row in sub[sub["is_outlier"]].iterrows():
                    all_out.append((label, row["w1"], row["w2"], row["w3"]))
                    ax.scatter(row["w1"], row["w2"], row["w3"],
                               s=300, facecolors="none", edgecolors="red",
                               linewidths=2, zorder=3)
                    ax.scatter(row["w1"], row["w2"], row["w3"],
                               s=50, color=color, edgecolor="black",
                               linewidth=0.5, zorder=6)

            print("Outliers for", key)
            for l,x,y,z in all_out:
                print(f"  {l}: x={x:.3f}, y={y:.3f}, z={z:.3f}")

            # Legend
            handles, labels = ax.get_legend_handles_labels()
            uniq = {}
            for h,l in zip(handles, labels):
                if l not in uniq: uniq[l] = h
            handles = list(uniq.values())
            labels  = list(uniq.keys())

            out_h = mlines.Line2D([],[], marker='o', color='none',
                                  markerfacecolor='none', markeredgecolor='red',
                                  markersize=10)
            handles.append(out_h)
            labels.append(f"Outliers (n={len(all_out)})")

            ax.legend(handles, labels, title="Label", loc='upper left')
            ax.set_xlim(xlims); ax.set_ylim(ylims)
            if zlims: ax.set_zlim(zlims)
            ax.set_xlabel("w1"); ax.set_ylabel("w2"); ax.set_zlabel("w3")
            ax.set_title(f"3D Gaussian Ellipsoid + Outliers – {key}")
            plt.tight_layout()
            plt.close(fig)
            #plt.show()

plot_weights_with_kde_and_outliers(raw_weights, colours_dict_label)
```

```python
def generate_synthetic_weights(real_data, model):
    m_factor = 2
    synthetic_weights = {}
    eval_metrics = {}

    for key, weights in real_data.items():
        is2d = weights.shape[1] == 3
        temp_synthetic_weights = []

        weights_data = weights[['w1', 'w2', 'true_label']] if is2d else weights[['w1', 'w2', 'w3', 'true_label']]
        log_l_list = []

        for label in weights_data['true_label'].unique():
            group = weights_data[weights_data['true_label'] == label][['w1', 'w2']] if is2d else \
                    weights_data[weights_data['true_label'] == label][['w1', 'w2', 'w3']]

            group = group.values
            try:
                if group.shape[0] < model.n_components:
                    continue
            except:
                pass

            num_cases = group.shape[0]
            model.fit(group)

            # Sample until we have enough strictly positive samples
            accepted = []
            while len(accepted) < num_cases * m_factor:
                needed = num_cases * m_factor - len(accepted)
                sampled = model.sample(needed * 2)[0] if isinstance(model, GaussianMixture) else model.sample(needed * 2)

                strictly_positive = sampled[(sampled > 0).all(axis=1)]
                accepted.extend(strictly_positive[:needed])

            synth_weight = np.array(accepted[:num_cases * m_factor])
            temp_data_pt = pd.DataFrame(synth_weight, columns=['w1', 'w2'] if is2d else ['w1', 'w2', 'w3'])
            temp_data_pt['true_label'] = label
            temp_synthetic_weights.append(temp_data_pt)

            if not isinstance(model, GaussianMixture):
                log_l = model.score_samples(group)
                log_l_list.extend(log_l)

        synthetic_weights[key] = pd.concat(temp_synthetic_weights, ignore_index=True)

        if isinstance(model, GaussianMixture):
            eval_metrics[key] = f'{model.bic(group):.4f}'
        else:
            n = len(group)
            k = group.shape[1]
            ll = np.sum(log_l_list)
            aic = 2 * k - 2 * ll
            bic = np.log(n) * k - 2 * ll
            eval_metrics[key] = f'{bic:.4f}'

    return synthetic_weights, eval_metrics
```

L2 Normalise the raw data

```python
def l2_normalise_raw_frames(raw_data_frames):
    """
    Takes the existing dict of raw_data_frames (run_key → DataFrame)
    and returns a new dict where each spectrum (row) is L2-normalised.
    
    Compatibility:
      - Keeps 'labels' column unchanged
      - Normalises only spectral intensity columns
      - Output DataFrames match the structure expected by your plotting functions
    """
    normalised_frames = {}

    for run_key, df in raw_data_frames.items():

        # Separate spectra and labels
        spectra_only = df.drop(columns=['labels']).values.astype(float)
        labels = df['labels'].copy()

        # Compute L2 norm per row
        norms = np.linalg.norm(spectra_only, axis=1, keepdims=True)

        # Avoid division by zero (flat spectra)
        norms[norms == 0] = 1.0

        # Normalise each spectrum
        spectra_norm = spectra_only / norms

        # Rebuild DataFrame
        norm_df = pd.DataFrame(spectra_norm, columns=df.columns[:-1])
        norm_df['labels'] = labels.values

        normalised_frames[run_key] = norm_df

    return normalised_frames

def l2_normalise_synthetic(data_array):
    """
    data_array: ndarray of shape (N_samples, N_points)
    Returns L2-normalised spectra (each row independently).
    """
    data = data_array.astype(float)
    norms = np.linalg.norm(data, axis=1, keepdims=True)
    norms[norms == 0] = 1.0   # avoid division by zero
    return data / norms

normalised_raw_data_frames = l2_normalise_raw_frames(raw_data_frames)
```

```python
def mean_squared_error(y_true, y_pred):
    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)
    return np.mean((y_true - y_pred)**2)

raw_vs_reconstructed_mse = {}  #

for key in normalised_raw_data_frames.keys():
    raw = normalised_raw_data_frames[key]

    if key in reconstructed_df.get('GMM', {}) and reconstructed_df['GMM'][key]:
        model_name = 'GMM'
        param_key = sorted(reconstructed_df['GMM'][key].keys())[-1]
        reconstructed_data = reconstructed_df[model_name][key][param_key][0]
    elif key in reconstructed_df.get('KDE', {}) and reconstructed_df['KDE'][key]:
        model_name = 'KDE'
        param_key = next(iter(reconstructed_df['KDE'][key].keys())) # Use first KDE result
        reconstructed_data = reconstructed_df[model_name][key][param_key][0]
    else:
        print(f"Skipping {key}: No reconstructed data found.")
        continue

    # Raw Data: Separate spectra and labels
    raw_spectra = raw.iloc[:, :-1].values
    raw_labels = raw['labels'].values
    
    recon_spectra = reconstructed_data['data']
    recon_labels = reconstructed_data['labels']

    experiment_mse = {}
    
    # Iterate over unique labels present in the raw data
    for label in sorted(np.unique(raw_labels)):
        # Mean raw spectrum for this label
        raw_mean = np.mean(raw_spectra[raw_labels == label], axis=0)

        # Get the subset of reconstructed samples with this label
        recon_subset = recon_spectra[recon_labels == label]
        
        if recon_subset.shape[0] == 0:
            continue

        # Compute MSEs: one MSE per reconstructed sample vs the raw mean reference
        mses = [mean_squared_error(recon_spectrum, raw_mean) for recon_spectrum in recon_subset]
        
        # Average MSE for this label
        experiment_mse[label] = np.mean(mses)

    raw_vs_reconstructed_mse[key] = experiment_mse

# Display results
print("\n--- Reference MSE (Raw Mean vs. Reconstructed Spectra) ---")
for key, label_mses in raw_vs_reconstructed_mse.items():
    print(f"\n{key}")
    for label, mse in label_mses.items():
        print(f"  {label}: {mse:.6f}")
```

```python
def compute_global_limits_for_model(
    model,
    raw_weights,
    raw_data_frames,
    gmm_weight_relabelled,
    kde_weight_relabelled,
    reconstructed_df,
    optimal_config_map,
):
    def update_limits(cur_min, cur_max, values):
        if values is None:
            return cur_min, cur_max
        arr = np.asarray(values).astype(float).ravel()
        arr = arr[np.isfinite(arr)]
        if arr.size == 0:
            return cur_min, cur_max
        mn, mx = arr.min(), arr.max()
        cur_min = mn if cur_min is None else min(cur_min, mn)
        cur_max = mx if cur_max is None else max(cur_max, mx)
        return cur_min, cur_max

    def finalize(cur_min, cur_max, pad=0.05, default=(0.0, 1.0)):
        if cur_min is None or cur_max is None:
            return default
        if not (np.isfinite(cur_min) and np.isfinite(cur_max)):
            return default
        span = cur_max - cur_min
        cur_min -= pad * span
        cur_max += pad * span
        return cur_min, cur_max

    w1_min = w1_max = None
    w2_min = w2_max = None
    w3_min = w3_max = None
    raw_y_min = raw_y_max = None
    other_y_min = other_y_max = None

    for key in raw_weights.keys():
        rw = raw_weights[key]
        raw_df = raw_data_frames[key]
        spec_cols = [c for c in raw_df.columns if c != "labels"]

        if "w1" in rw: w1_min, w1_max = update_limits(w1_min, w1_max, rw["w1"])
        if "w2" in rw: w2_min, w2_max = update_limits(w2_min, w2_max, rw["w2"])
        if "w3" in rw: w3_min, w3_max = update_limits(w3_min, w3_max, rw["w3"])
        raw_y_min, raw_y_max = update_limits(raw_y_min, raw_y_max, raw_df[spec_cols].values)

        optimal_param = optimal_config_map[model].get(key)
        if optimal_param is None:
            continue

        if model == "GMM":
            weights_dict = gmm_weight_relabelled
            model_key = "GMM"
        else:
            weights_dict = kde_weight_relabelled
            model_key = "KDE"

        try:
            other_w = weights_dict[key][optimal_param]
            entry = reconstructed_df[model_key][key][optimal_param][0]
        except KeyError:
            continue

        if "w1" in other_w: w1_min, w1_max = update_limits(w1_min, w1_max, other_w["w1"])
        if "w2" in other_w: w2_min, w2_max = update_limits(w2_min, w2_max, other_w["w2"])
        if "w3" in other_w: w3_min, w3_max = update_limits(w3_min, w3_max, other_w["w3"])
        other_y_min, other_y_max = update_limits(other_y_min, other_y_max, entry["data"])

    # Use raw_y if other_y never got set
    if other_y_min is None or other_y_max is None:
        other_y_min, other_y_max = raw_y_min, raw_y_max
    # Final fallbacks
    if raw_y_min is None or raw_y_max is None:
        raw_y_min, raw_y_max = 0.0, 1.0
    if other_y_min is None or other_y_max is None:
        other_y_min, other_y_max = 0.0, 1.0

    return {
        "w1": finalize(w1_min, w1_max),
        "w2": finalize(w2_min, w2_max),
        "w3": finalize(w3_min, w3_max),
        "raw_y": finalize(raw_y_min, raw_y_max, pad=0.0, default=(0.0, 1.0)),
        "other_y": finalize(other_y_min, other_y_max, pad=0.0, default=(0.0, 1.0)),
    }

def compute_shading(weights_df, colours_dict_label):
    """Assigns consistent shading for all plots (used in scatter + spectra)."""
    w_cols = [c for c in weights_df.columns if c.startswith("w")]
    d = np.linalg.norm(weights_df[w_cols].values, axis=1)
    dmin, dmax = d.min(), d.max()

    shades = [
        shade_by_distance(colours_dict_label.get(lbl, "black"), dist, dmin, dmax)
        for lbl, dist in zip(weights_df["true_label"], d)
    ]
    weights_df["shade"] = shades
    return weights_df


# ======================================================================
# SCATTER HELPERS (FIXED)
# ======================================================================

def plot_simple_scatter(ax, df, colours_dict_label, global_limits):
    """Simple 2D/3D scatter for OTHER weights (no SD, no outliers).
    FIX: Proper color mapping via shading column for both 2D and 3D."""
    is_3d = ("w3" in df.columns)

    if not is_3d:
        # 2D case
        for lab in sorted(df["true_label"].unique()):
            sub = df[df["true_label"] == lab]
            # Use 'shade' column for colors
            ax.scatter(
                sub["w1"], sub["w2"],
                c=list(sub["shade"]),  # FIXED: use shading
                s=45, edgecolor="k", linewidth=0.3,
                label=lab
            )
        ax.set_xlabel("w1")
        ax.set_ylabel("w2")
        ax.set_xlim(*global_limits["w1"])
        ax.set_ylim(*global_limits["w2"])
        ax.grid(True)  # FIX 1: Add grid
        ax.legend(title="Label")
    else:
        # 3D case
        for lab in sorted(df["true_label"].unique()):
            sub = df[df["true_label"] == lab]
            # Use 'shade' column for colors
            ax.scatter(
                sub["w1"], sub["w2"], sub["w3"],
                c=list(sub["shade"]),  # FIXED: use shading
                s=45, edgecolor="k", linewidth=0.3,
                label=lab
            )
        ax.set_xlabel("w1")
        ax.set_ylabel("w2")
        ax.set_zlabel("w3")
        ax.set_xlim(*global_limits["w1"])
        ax.set_ylim(*global_limits["w2"])
        if global_limits["w3"][0] is not None:
            ax.set_zlim(*global_limits["w3"])
        ax.legend(title="Label")


def plot_advanced_scatter(ax, df, colours_dict_label, global_limits, outlier_sd=3.0, gridsize=120, use_kde_countours=False):
    """
    Advanced scatter:
      - 2D: Gaussian SD rings (1/2/3σ), KDE contours, z-score outliers.
      - 3D: Gaussian ellipsoids (1/2/3σ) + z-score outliers.
    FIX 4: Outliers are now plotted with visible markers, not filtered.
    """
    is_3d = ("w3" in df.columns)

    # ---------- 2D ----------
    if not is_3d:
        xlims = global_limits["w1"]
        ylims = global_limits["w2"]

        xg = np.linspace(*xlims, gridsize)
        yg = np.linspace(*ylims, gridsize)
        X, Y = np.meshgrid(xg, yg)
        pts = np.column_stack([X.ravel(), Y.ravel()])

        all_out = []

        for lab in sorted(df["true_label"].unique()):
            sub = df[df["true_label"] == lab].copy()
            if sub.empty:
                continue

            base_color = colours_dict_label.get(lab, "C0")
            data = sub[["w1", "w2"]].astype(float).to_numpy()

            # KDE contours
            if len(sub) > 3:
                if use_kde_countours:
                    try:
                        kde = gaussian_kde(data.T)
                        Z = kde(pts.T).reshape(X.shape)
                        ax.contour(X, Y, Z, levels=5, colors=[base_color], alpha=0.5)
                    except Exception:
                        pass

                # Gaussian SD rings
                mean = data.mean(axis=0)
                cov = np.cov(data, rowvar=False) + np.eye(2) * 1e-6
                cov_inv = np.linalg.pinv(cov)

                diff = pts - mean
                d2 = np.einsum("ij,jk,ik->i", diff, cov_inv, diff).reshape(X.shape)
                D = np.sqrt(d2)

                alphas = [0.45, 0.25, 0.12]
                for i in range(3):
                    col = np.array(mcolors.to_rgba(base_color))
                    col[3] = alphas[i]
                    ax.contourf(
                        X, Y, D,
                        levels=[i, i + 1],
                        colors=[col],
                        antialiased=True
                    )

                ax.contour(X, Y, D, levels=[1, 2, 3],
                           colors=[base_color], linewidths=1.2)

            # Plot ALL inlier points with shading
            ax.scatter(
                sub["w1"], sub["w2"],
                c=list(sub["shade"]),
                s=45, edgecolor="k", linewidth=0.3
            )

        # FIX 4: Plot outliers SEPARATELY with clear markers
        for lab in sorted(df["true_label"].unique()):
            sub = df[df["true_label"] == lab].copy()
            if sub.empty:
                continue
            
            # Compute outliers
            zs = (sub[["w1","w2"]] - sub[["w1","w2"]].mean()) / sub[["w1","w2"]].std().replace(0,1)
            out_mask = (zs.abs() > outlier_sd).any(axis=1)         
            
            for idx, row in sub.loc[out_mask].iterrows():
                all_out.append((lab, row["w1"], row["w2"]))
                # Large red unfilled circle for visibility
                ax.scatter(
                    row["w1"], row["w2"],
                    s=40, c=colours_dict_label.get(lab, "k"),
                    edgecolors="k", linewidths=0.5, zorder=9
                )

                ax.scatter(
                    row["w1"], row["w2"],
                    s=200, facecolors="none",
                    edgecolors="red", linewidths=2.5, zorder=10
                )

        ax.set_xlim(*xlims)
        ax.set_ylim(*ylims)
        ax.set_xlabel("w1")
        ax.set_ylabel("w2")
        ax.grid(True)  # FIX 1: Add grid

        # Legend
        legend_handles = []
        legend_labels  = []

        # 1) class handles
        for lab in sorted(df["true_label"].unique()):
            legend_handles.append(
                mlines.Line2D([], [], marker="o",
                            color=colours_dict_label.get(lab,"C0"),
                            markersize=8, linestyle="None")
            )
            legend_labels.append(lab)

        # 2) outlier entry (with count)
        out_count = len(all_out)
        legend_handles.append(
            mlines.Line2D([], [], marker="o", color="none",
                        markerfacecolor="none", markeredgecolor="red",
                        markersize=10, linewidth=2.5)
        )
        legend_labels.append(f"Outliers")# (n={out_count})")

        ax.legend(legend_handles, legend_labels, title="Label", loc="upper left")

    # ---------- 3D ----------
    else:
        xlims = global_limits["w1"]
        ylims = global_limits["w2"]
        zlims = global_limits["w3"]

        all_out = []
        
        # First pass: Plot inliers and ellipsoids for all labels
        for lab in sorted(df["true_label"].unique()):
            sub = df[df["true_label"] == lab].copy()
            if sub.empty:
                continue

            base_color = colours_dict_label.get(lab, "C0")

            mean_global = df[["w1","w2","w3"]].mean()
            std_global  = df[["w1","w2","w3"]].std().replace(0,1)
            zs_full = np.abs((sub[["w1","w2","w3"]] - mean_global) / std_global)
            out_mask = (zs_full > outlier_sd).any(axis=1)
            inliers = sub.loc[~out_mask, ["w1", "w2", "w3"]].astype(float).to_numpy()

            if len(inliers) >= 5:
                mean3 = inliers.mean(axis=0)
                cov3 = np.cov(inliers, rowvar=False)
                for n_std, alpha in zip([1, 2, 3], [0.30, 0.18, 0.10]):
                    plot_gaussian_ellipsoid(ax, mean3, cov3, n_std=n_std,
                                            color=base_color, alpha=alpha)

            ax.scatter(
                sub.loc[~out_mask, "w1"],
                sub.loc[~out_mask, "w2"],
                sub.loc[~out_mask, "w3"],
                c=list(sub.loc[~out_mask, "shade"]),
                s=40, label=lab, zorder=5
            )
        
        # Second pass: Plot ONLY outliers last (so they're always on top)
        for lab in sorted(df["true_label"].unique()):
            sub = df[df["true_label"] == lab].copy()
            if sub.empty:
                continue

            zs = (sub[["w1","w2","w3"]] - sub[["w1","w2","w3"]].mean()) / sub[["w1","w2","w3"]].std().replace(0,1)
            out_mask = (zs.abs() > outlier_sd).any(axis=1)

            for _, row in sub.loc[out_mask].iterrows():
                all_out.append((lab, row["w1"], row["w2"], row["w3"]))
                # Large red unfilled circle - drawn LAST so it's on top
                ax.scatter(
                    row["w1"], row["w2"], row["w3"],
                    s=40, c=colours_dict_label.get(lab, "k"),
                    edgecolors="k", linewidths=0.5, zorder=9
                )

                # plot red circle around it
                ax.scatter(
                    row["w1"], row["w2"], row["w3"],
                    s=300, facecolors="none", edgecolors="red",
                    linewidths=2.5, zorder=10
                )

        ax.set_xlim(*xlims)
        ax.set_ylim(*ylims)
        if zlims[0] is not None:
            ax.set_zlim(*zlims)
        ax.set_xlabel("w1")
        ax.set_ylabel("w2")
        ax.set_zlabel("w3")

        handles, labels = ax.get_legend_handles_labels()
        uniq = {}
        for h, l in zip(handles, labels):
            if l not in uniq:
                uniq[l] = h
        handles = list(uniq.values())
        labels = list(uniq.keys())
        out_h = mlines.Line2D([], [], marker="o", color="none",
                              markerfacecolor="none", markeredgecolor="red",
                              markersize=10, linewidth=2.5)
        handles.append(out_h)
        labels.append(f"Outliers (n={len(all_out)})")
        ax.legend(handles, labels, title="Label", loc="upper left") 
        ax.legend(handles, labels, title="Label", loc="upper left")


# ======================================================================
# FULL PAGE: 2 ROWS × (1 + N_CLASSES) COLUMNS (FIXED)
# ======================================================================
def plot_full_page(
    real_weights_df,
    raw_df,
    other_weights_df,
    other_spectra_data,
    other_labels,
    colours_dict_label,
    global_limits,
    ppm_axis=None,  # FIX 2: Add PPM axis parameter
    outlier_sd=3.0,
    gridsize=120,
    use_kde_countours=False,
    experiment_name='Test'
):
    """
    Layout for n classes:

      ROW 0: [ advanced scatter (REAL weights) | RAW spectra per class ]
      ROW 1: [ simple scatter (OTHER weights)  | OTHER spectra per class ]
    
    FIX 2: PPM axis (4.5 to 0) used for all spectra x-axes
    """

    real_weights_df = real_weights_df.copy()
    other_weights_df = other_weights_df.copy()

    if "true_label" not in real_weights_df.columns and "label" in real_weights_df.columns:
        real_weights_df["true_label"] = real_weights_df["label"]
    if "true_label" not in other_weights_df.columns and "label" in other_weights_df.columns:
        other_weights_df["true_label"] = other_weights_df["label"]

    real_weights_df["true_label"] = real_weights_df["true_label"].astype(str)
    other_weights_df["true_label"] = other_weights_df["true_label"].astype(str)

    real_weights_df = compute_shading(real_weights_df, colours_dict_label)
    other_weights_df = compute_shading(other_weights_df, colours_dict_label)

    unique_labels = sorted(real_weights_df["true_label"].unique())
    n_classes = len(unique_labels)

    # Build OTHER spectra DataFrame
    spec_cols = [c for c in raw_df.columns if c != "labels"]
    other_spectra_df = pd.DataFrame(other_spectra_data, columns=spec_cols)
    other_spectra_df["labels"] = other_labels

    fig = plt.figure(figsize=(10 + 6*n_classes, 6 + 5*n_classes), dpi=1_00)
    gs = fig.add_gridspec(
        nrows=2,
        ncols=1 + n_classes,
        width_ratios=[2.5] + [2]*n_classes,
        height_ratios=[1, 1],
        hspace=0.28,
        wspace=0.25,
    )

    # FIX 2: PPM axis setup
    if ppm_axis is None:
        n_points = len(spec_cols)
        ppm_axis = np.linspace(4.5, 0, n_points)
    
    # ---------- ROW 0, COL 0: ADVANCED REAL SCATTER ----------
    ax0 = fig.add_subplot(gs[0, 0], projection="3d" if "w3" in real_weights_df.columns else None)
    plot_advanced_scatter(
        ax0,
        real_weights_df,
        colours_dict_label,
        global_limits,
        outlier_sd=outlier_sd,
        gridsize=gridsize,
        use_kde_countours=use_kde_countours
    )
    ax0.set_title("REAL weights: SD rings + outliers")

    # ---------- ROW 0, COL 1..N: RAW SPECTRA PER CLASS ----------
    combined_ymin = min(global_limits["raw_y"][0], global_limits["other_y"][0])
    combined_ymax = max(global_limits["raw_y"][1], global_limits["other_y"][1])
    ybuf = 0.01 * (combined_ymax - combined_ymin)
    combined_ymin -= ybuf
    combined_ymax += ybuf

    for i, lab in enumerate(unique_labels):
        ax = fig.add_subplot(gs[0, 1 + i])
        sub_raw = raw_df[raw_df["labels"] == lab]

        for idx in sub_raw.index:
            if idx < len(real_weights_df):
                col = real_weights_df["shade"].iloc[idx]
            else:
                col = colours_dict_label.get(lab, "k")
            # FIX 2: Use PPM axis
            ax.plot(ppm_axis, sub_raw.loc[idx, spec_cols].values, color=col, alpha=0.8)

        ax.set_title(f"Raw – {lab}")
        ax.set_xlabel("Chemical Shift (ppm)")
        ax.set_ylabel("Intensity")
        ax.set_xlim(4.5, 0)  # ppm goes from high to low
        ax.set_ylim(combined_ymin, combined_ymax)
        ax.grid(True)  # FIX 1: Grid added

    # ---------- ROW 1, COL 0: SIMPLE OTHER SCATTER ----------
    ax1 = fig.add_subplot(gs[1, 0], projection="3d" if "w3" in other_weights_df.columns else None)
    plot_simple_scatter(ax1, other_weights_df, colours_dict_label, global_limits)
    ax1.set_title("SYNTHETIC weights")

    # ---------- ROW 1, COL 1..N: OTHER SPECTRA PER CLASS ----------
    for i, lab in enumerate(unique_labels):
        ax = fig.add_subplot(gs[1, 1 + i])
        sub_other = other_spectra_df[other_spectra_df["labels"] == lab]

        for idx in sub_other.index:
            if idx < len(other_weights_df):
                col = other_weights_df["shade"].iloc[idx]
            else:
                col = colours_dict_label.get(lab, "k")
            # FIX 2: Use PPM axis
            ax.plot(ppm_axis, sub_other.loc[idx, spec_cols].values, color=col, alpha=0.8)

        ax.set_title(f"Synthetic – {lab}")
        ax.set_xlabel("Chemical Shift (ppm)")
        ax.set_ylabel("Intensity")
        ax.set_xlim(4.5, 0)  # ppm goes from high to low
        ax.set_ylim(combined_ymin, combined_ymax)
        ax.grid(True)  # FIX 1: Grid added

    fig.suptitle(f"{model} – {experiment_name}: Real vs Synthetic Weights & Spectra", fontsize=16)
    plt.tight_layout()
    plt.savefig(f'{nb_dir}/img/{model}/{experiment_name}.png', dpi=600, bbox_inches='tight', transparent=True)
    plt.close(fig)
```

```python
def parse_mrs_xml(filepath):
    tree = ET.parse(filepath)
    root = tree.getroot()
    
    records = []
    
    for case in root.findall('Case'):
        case_id = case.get('ID')
        is_synthetic = case_id.startswith('SYN')
        
        tissue_type = case.find('Tissue').get('Type').lower()
        spectrum = list(map(float, case.find('.//Points').text.strip().split()))
        
        records.append({
            'id': case_id,
            'label': tissue_type,
            'synthetic': is_synthetic,
            'spectrum': np.array(spectrum)
        })

    return pd.DataFrame(records)

def crop_real_spectrum_to_synthetic_range(spectrum, 
                                           real_N=512, real_min=-2.7, real_max=7.1,
                                           target_min=0.0, target_max=4.5):
    # Compute indices in real spectrum corresponding to target range
    start_idx = get_PPM(target_max, real_N, real_max, real_min)
    end_idx   = get_PPM(target_min, real_N, real_max, real_min)
    
    # Slice and flip (if needed)
    cropped = spectrum[start_idx:end_idx + 1]  # +1 to include endpoint
    return cropped
```

```python
data_cropped = {}

for model in data:
    data_cropped[model] = {}
    for exp in data[model]:
        data_cropped[model][exp] = {}
        for te in data[model][exp]:
            df = data[model][exp][te].copy()

            for i, row in df.iterrows():
                if not row['synthetic']:
                    cropped = crop_real_spectrum_to_synthetic_range(row['spectrum'])
                    df.at[i, 'spectrum'] = cropped  # overwrite spectrum with cropped version

            data_cropped[model][exp][te] = df
```
